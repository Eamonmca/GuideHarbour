# Transport-free audit service

This layer is deliberately JSON-in/JSON-out, so a CLI, REST application, or
MCP server can adapt it without embedding scientific logic.

```
#| default_exp service
```

```
#| export
from __future__ import annotations

from typing import Any, Mapping

from probeharbour.records import audit_candidate
from probeharbour.rules import RuleSet
```

```
#| export
class AuditService:
    """JSON-safe operations intended for CLI, API, and MCP adapters."""

    def __init__(self, ruleset: RuleSet, candidates: Mapping[str, Mapping[str, Any]]):
        self.ruleset = ruleset
        self.candidates = dict(candidates)

    def get_candidate(self, candidate_id: str) -> dict[str, Any]:
        return dict(self.candidates[candidate_id])

    def audit_candidate(self, candidate_id: str) -> dict[str, Any]:
        return audit_candidate(self.candidates[candidate_id])

    def explain_score(self, candidate_id: str) -> dict[str, Any]:
        candidate = self.candidates[candidate_id]
        result = self.ruleset.explain(candidate).as_dict()
        result["stored_raw_design_score"] = candidate.get("stored_raw_design_score", candidate.get("raw_design_score"))
        return result
```
