# ProbeStore audit contracts

This nbdev-ready notebook is the human-readable home for the record and score contracts. Implementation starts in the package while the contracts are still being stabilised.

## Non-negotiable first contracts

- `stored_raw_design_score` is immutable source evidence.
- `recalculated_score` always names its ruleset.
- `on_junction` and `junction_spanning` are separate booleans.
- A spacing violation requires nearest-probe name and distance evidence.
- MCP transports plain JSON to and from `AuditService`.
