```
#| export
"""Small public, compatibility-first API for ProbeHarbour."""
from __future__ import annotations

from probeharbour.design import generate_junction_candidates, select_candidates
from probeharbour.models import Provenance
from probeharbour.rules import RuleSet
from probeharbour.specificity import SpecificityService
```

```
#| export
def load_rules(chemistry: str) -> RuleSet:
    """Load the active chemistry rules (`prime_5k` or `xenium_v1`)."""
    return RuleSet.from_packaged(chemistry)
```

```
#| export
def design_junction(sequence: str, junction_position: int, *, chemistry: str = "prime_5k", arm_length: int = 16,
                    slides=range(-5, 6), candidate_prefix: str = "candidate", dataset_id: str = "interactive"):
    """Generate the familiar slide-based candidate set around one junction."""
    return generate_junction_candidates(sequence, junction_position=junction_position, arm_length=arm_length,
        slides=slides, provenance=Provenance("designed", dataset_id, ruleset_id=load_rules(chemistry).ruleset_id), candidate_prefix=candidate_prefix)
```

```
#| export
def select(rows, *, n_best: int = 1):
    """Select best candidate(s) while retaining decisions for every row."""
    return select_candidates(rows, limit=n_best)
```
