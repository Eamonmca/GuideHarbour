# Sequence facts and QC

These helpers are pure functions: they can be used by design, imports,
dashboard record cards, and tests without loading Primer3 or transcript data.

```
#| default_exp sequence
```

```
#| export
from __future__ import annotations

import re

SIMPLE_REPEAT_RE = re.compile(r"([ACGT]{3,4})\1{2,}", re.IGNORECASE)
# A C-run in the probe binds a G-run in the target RNA.
FIVE_G_RE = re.compile(r"C{5,}", re.IGNORECASE)
```

```
#| export
def normalize_sequence(sequence: str) -> str:
    value = str(sequence).upper().strip()
    if not value or any(base not in "ACGT" for base in value):
        raise ValueError("Sequence must be non-empty and contain only A/C/G/T")
    return value
```

```
#| export
def ligation_dinucleotide(rbd5: str, rbd3: str) -> str:
    return normalize_sequence(rbd5)[-1] + normalize_sequence(rbd3)[0]
```

```
#| export
def gc_percent(sequence: str) -> float:
    value = normalize_sequence(sequence)
    return round(100 * (value.count("G") + value.count("C")) / len(value), 3)
```

```
#| export
def has_simple_repeat(sequence: str) -> bool:
    return bool(SIMPLE_REPEAT_RE.search(normalize_sequence(sequence)))
```

```
#| export
def has_5g_repeat(rbd5: str, rbd3: str) -> bool:
    return bool(FIVE_G_RE.search(normalize_sequence(rbd5)) or FIVE_G_RE.search(normalize_sequence(rbd3)))
```
